# 项目：评估和清理英国电商公司销售数据

## 分析目标

此数据分析的目的是，根据市场销售数据，挖掘畅销产品，以便制定更有效的市场策略来提升营收。

本实战项目的目的在于练习评估数据干净和整洁度，并且基于评估结果，对数据进行清洗，从而得到可供下一步分析的数据。

## 简介

原始数据集记录了一家英国在线零售公司在2010年12月1日至2011年12月9日期间的所有交易情况，涵盖了该公司在全球不同国家和地区的业务数据。该公司主要销售覆盖各个场景的礼品，包括但不限于生日礼品、结婚纪念品、圣诞礼品等等。该公司的客户群体主要包括批发商和个人消费者，其中批发商占据了相当大的比例。

数据每列的含义如下：
- `InvoiceNo`: 发票号码。6位数，作为交易的唯一标识符。如果这个代码以字母“c”开头，表示这笔交易被取消。
- `StockCode`: 产品代码。5位数，作为产品的唯一标识符。
- `Description`: 产品名称。
- `Quantity`: 产品在交易中的数量。
- `InvoiceDate`: 发票日期和时间。交易发生的日期和时间。
- `UnitPrice`: 单价。价格单位为英镑（£）。
- `CustomerID`: 客户编号。5位数，作为客户的唯一标识符。
- `Country`: 国家名称。客户所居住的国家的名称。

## 读取数据

导入数据分析所需要的库，并通过pandas的read_csv函数，将原始数据文件“e_commerce.csv”里面的数据内容，解析为DataFrame，并且赋值给变量original_data

```python
import pandas as pd
```

```python
original_data = pd.read_csv("e_commerce.csv")
```

```python
original_data.head()
```

## 评估数据

对上一部分建立的original_data这个DataFrame所包含的数据进行评估。
评估主要从两个方面进行：结构和内容，即整齐度和干净度。
数据的结构性问题指不符合“每行一个观察值，每列一个变量，每个单元格是一个值”这三个标准，数据的内容性问题包括存在丢失数据、重复数据、无效数据等。

### 评估数据整齐度

```python
original_data.sample(10)
```

数据符合“每行是一个观察值，每列是一个变量，每个单元格是一个值”的标准，每行是关于某商品的一次交易，每列是交易相关的各个变量，因此不存在结构性问题

### 评估数据干净度

```python
original_data.info()
```

数据共有541909条观察值，而`Description`、`CustomerID`具有缺失值。
`InvoiceDate`的数据类型应该为日期，`CustomerID`的数据类型应该为字符串，应该进行数据格式转化。

#### 评估缺失数据

```python
original_data[original_data["Description"].isnull()]
```

```python
original_data[(original_data["Description"].isnull()) & (original_data["UnitPrice"] != 0)]
```

筛选出来结果数量为0条，说明缺失`Description`的数据同时也不具备有效的`UnitPrice`值。
`Description`表示产品名称，`UnitPrice`表示产品单价，都是后续进行商品交易分析的重要变量。如果他们同时缺失/无效，我们认为数据无法提供有效含义，因此可以后续删除。

```python
original_data[original_data["CustomerID"].isnull()]
```

`CustomerID`表示客户编号，不是分析畅销商品的必要变量，所以这些数据仍然有效，因此保留此变量为空的观察值。

#### 评估重复数据

虽然`InvoiceNo`、`StockCode`和`CustomerID`都是唯一标识符，但一次交易可能包含多件商品，所以`InvoiceNo`可以重复；不同交易可以包含同一件商品，所以`StockCode`可以重复；顾客可以多次交易或者下单多个商品，所以`CustomerID`可以重复。
因此，针对此数据我们无需评估重复数据。

#### 评估不一致数据

`Country`变量中，我们要查看是否存在多个不同值指代同一个国家的情况。

```python
original_data["Country"].value_counts().sort_index()
```

`USA`、`United States`都在表示美国，`UK`、`U.K.`、`United Kingdom`都在表示英国，应该对其统一只保留一个指代值。

#### 评估无效或错误数据

```python
original_data.describe()
```

从输出结果来看，`Quantity`、`UnitPrice`存在负数。
应该筛选出来进一步评估其含义。

```python
original_data[original_data["Quantity"] < 0]
```

```python
original_data[(original_data["Quantity"] < 0) & (original_data["InvoiceNo"].str[0] != "C")]
```

```python
original_data[(original_data["Quantity"] < 0) & (original_data["InvoiceNo"].str[0] != "C") & (original_data["UnitPrice"] != 0)]
```

猜想得到验证，当`Quantity`变量为负数的时候，观察值满足以下条件之一：  
- `InvoiceNo`以C开头，表示订单被取消
- `UnitPrice`为0，说明单价为0英镑
   
这些交易数据都不是有效成交数据，不在后续分析范围之内，所以我们将在数据清理步骤里将`Quantity`为负数的观察值删除。

```python
original_data[original_data["UnitPrice"] < 0]
```

`UnitPrice`为负数的都是坏账调整，不属于实际商品交易数据，在数据清理步骤中也将其删除。

## 清洗数据

根据前面评估部分得到的结论，需要进行的数据清理包括：  
- `InvoiceDate`的数据类型转化为日期
- `CustomerID`的数据类型转化为字符串
- `Description`变量缺失的观察值删除
- 把`Country`变量值"USA"替换为"United States"
- 把`Country`变量值"UK"、"U.K."替换成"United Kingdom"
- 将`Quantity`为负数的观察值删除
- 将`UnitPrice`为负数的观察值删除

```python
cleaned_data = original_data.copy()
cleaned_data.head()
```

`InvoiceDate`的数据类型转化为日期

```python
cleaned_data["InvoiceDate"] = pd.to_datetime(cleaned_data["InvoiceDate"])
cleaned_data["InvoiceDate"]
```

 `CustomerID`的数据类型转化为字符串

```python
cleaned_data["CustomerID"] = cleaned_data["CustomerID"].astype(str)
cleaned_data["CustomerID"]
```

 `CustomerID`的变量值结尾“.0”去掉

```python
cleaned_data["CustomerID"] = cleaned_data["CustomerID"].str.slice(0,-2)
cleaned_data["CustomerID"]
```

`Description`变量缺失的观察值删除

```python
cleaned_data.dropna(subset=["Description"],inplace=True)
```

```python
cleaned_data["Description"].isnull().sum()
```

把`Country`变量值"USA"替换为"United States"

```python
cleaned_data["Country"] = cleaned_data["Country"].replace({"USA":"United States"})
```

```python
len(cleaned_data[cleaned_data["Country"] == "USA"])
```

把`Country`变量值"UK"、"U.K."替换成"United Kingdom"

```python
cleaned_data["Country"] = cleaned_data["Country"].replace({"UK":"United Kingdom","U.K.":"United Kingdom"})
```

```python
len(cleaned_data[cleaned_data["Country"] == "UK"])
```

```python
len(cleaned_data[cleaned_data["Country"] == "U.K."])
```

将`Quantity`为负数的观察值删除

```python
cleaned_data = cleaned_data[cleaned_data["Quantity"] >= 0]
```

```python
len(cleaned_data[cleaned_data["Quantity"] < 0])
```

将`UnitPrice`为负数的观察值删除

```python
cleaned_data = cleaned_data[cleaned_data["UnitPrice"] >= 0]
```

```python
len(cleaned_data[cleaned_data["UnitPrice"] < 0])
```

## 保存清理后的数据

```python
cleaned_data.to_csv("e_commerce_cleaned.csv",index = False)
```
